```python
import joblib
import pandas as pd
import numpy as np
```

```python

model = joblib.load('anaemia_logistic_regression_model.pkl')

scaler = joblib.load('scaler.pkl')
```

```python
# Generating synthetic data for 50 rows
np.random.seed(42)  # for reproducibility

# Generating random data for each feature
data = {
    'Sex': np.random.choice([0, 1], size=50),  # 0 or 1 for sex
    '%Red Pixel': np.random.uniform(40, 50, size=50),  # random values between 40 and 50
    '%Green pixel': np.random.uniform(25, 35, size=50),  # random values between 25 and 35
    '%Blue pixel': np.random.uniform(20, 30, size=50),  # random values between 20 and 30
    'Hb': np.random.uniform(6, 15, size=50)  # random values between 6 and 15 for Hb
}

new_data = pd.DataFrame(data)
```

```python
print(new_data.head())
```

```python

new_data_scaled = scaler.transform(new_data)


predictions = model.predict(new_data_scaled)

print("Predictions:", predictions)
```

```python
#new df
result_df = pd.concat([new_data, pd.Series(predictions, name='Predictions')], axis=1)

# Display df
print(result_df)
```
